**1.The change of sperm number**

```python
import matplotlib.pyplot as plt
import numpy as np

# 数据（用 np.nan 替换 None）
time_points = [0, 2, 6, 12, 24, "24*", "48*"]

uterus_short = [ 261.4, 243.4, 238.3, 279.2, 531.4, 9.3, 0.8]
uterus_short_se = [5.6, 17.1, 10.5, 16.5, 99.1, 8.9, 0.3]

uterus_long = [285.6, 308.0, 290.5, 241.2, 380.0, 18.5, 19.1]
uterus_long_se = [6.5, 16.3, 9.4, 16.4, 69.6, 8.8, 7.4]

sp_short = [np.nan, 19.1, 211.4, 69.4, 60.0, 42.3, 47.9]
sp_short_se = [np.nan, 6.3, 55.1, 16.3, 16.8, 27.1, 8.0]

sp_long = [np.nan, 10.5, 492.7, 726.8, 1371.9, 1155.3, 1618.3]
sp_long_se = [np.nan, 5.2, 154.9, 191.6, 190.1, 320.7, 119.5]

# 分组索引
non_ovip_idx = [0,1,2,3,4]   # -1,0,2,6,12,24
ovip_idx = [5,6]               # 24*,48

# 画图
fig, axes = plt.subplots(1, 2, figsize=(4.5,2.5), sharey=True)

# ---------- Non-oviposited ----------
ax = axes[0]
x = np.arange(len(non_ovip_idx))

ax.errorbar(x, [uterus_short[i] for i in non_ovip_idx],
            yerr=[uterus_short_se[i] for i in non_ovip_idx],
            label="Uterus Shorter", color="#385B6F", linestyle="--", marker="o")

ax.errorbar(x, [uterus_long[i] for i in non_ovip_idx],
            yerr=[uterus_long_se[i] for i in non_ovip_idx],
            label="Uterus Long", color="#385B6F", linestyle="-", marker="o")

ax.errorbar(x, [sp_short[i] for i in non_ovip_idx],
            yerr=[sp_short_se[i] for i in non_ovip_idx],
            label="Spermatheca Shorter", color="#AA5440", linestyle="--", marker="s")

ax.errorbar(x, [sp_long[i] for i in non_ovip_idx],
            yerr=[sp_long_se[i] for i in non_ovip_idx],
            label="Spermatheca Long", color="#AA5440", linestyle="-", marker="s")

ax.set_xticks(x)
ax.set_xticklabels([time_points[i] for i in non_ovip_idx])
ax.set_title("Non-oviposited")
ax.set_xlabel("Hours post-mating (hr)")
ax.set_ylabel("Sperm number (mean ± SE)")

# ---------- Oviposited ----------
ax = axes[1]
x = np.arange(len(ovip_idx))

ax.errorbar(x, [uterus_short[i] for i in ovip_idx],
            yerr=[uterus_short_se[i] for i in ovip_idx],
            label="Uterus Shorter", color="#385B6F", linestyle="--", marker="o")

ax.errorbar(x, [uterus_long[i] for i in ovip_idx],
            yerr=[uterus_long_se[i] for i in ovip_idx],
            label="Uterus Long", color="#385B6F", linestyle="-", marker="o")

ax.errorbar(x, [sp_short[i] for i in ovip_idx],
            yerr=[sp_short_se[i] for i in ovip_idx],
            label="Spermatheca Shorter", color="#AA5440", linestyle="--", marker="s")

ax.errorbar(x, [sp_long[i] for i in ovip_idx],
            yerr=[sp_long_se[i] for i in ovip_idx],
            label="Spermatheca Long", color="#AA5440", linestyle="-", marker="s")

ax.set_xticks(x)
ax.set_xticklabels([time_points[i] for i in ovip_idx])
ax.set_title("Oviposited")
ax.set_xlabel("Hours post-mating (hr)")

# ---------- Legend ----------
fig.legend(loc="right", bbox_to_anchor=(0.5, 1.05), ncol=4)
plt.tight_layout()
#plt.savefig("D:/Documents/PMR/文献/2sperm/2sperm_number_change_PM.pdf", dpi=300, bbox_inches='tight', format='pdf')
plt.show()
```

**2. The change of sperm proportion**

```python
import matplotlib.pyplot as plt
import numpy as np

# 时间点
time_points = [0, 2, 6, 12, 24, "24*", "48*"]

# 数据
uterus_short = [261.4, 243.4, 238.3, 279.2, 531.4, 9.3, 0.8]
uterus_long = [285.6, 308.0, 290.5, 241.2, 380.0, 18.5, 19.1]

sp_short = [np.nan, 19.1, 211.4, 69.4, 60.0, 42.3, 47.9]
sp_long = [np.nan, 10.5, 492.7, 726.8, 1371.9, 1155.3, 1618.3]

# 计算占比
uterus_total = np.array(uterus_short) + np.array(uterus_long)
sp_total = np.array(sp_short) + np.array(sp_long)

uterus_short_ratio = np.array(uterus_short) / uterus_total
uterus_long_ratio = np.array(uterus_long) / uterus_total

sp_short_ratio = np.array(sp_short) / sp_total
sp_long_ratio = np.array(sp_long) / sp_total

# 分组索引
non_ovip_idx = [0,1,2,3,4]   # -1,0,2,6,12,24
ovip_idx = [5,6]               # 24*,48

# 画图
fig, axes = plt.subplots(1, 2, figsize=(4.5,2.5), sharey=True)

# ---------- Non-oviposited ----------
ax = axes[0]
x = np.arange(len(non_ovip_idx))

ax.plot(x, uterus_short_ratio[non_ovip_idx], label="Uterus Shorter", color="#385B6F", linestyle="--", marker="o")
ax.plot(x, uterus_long_ratio[non_ovip_idx], label="Uterus Long", color="#385B6F", linestyle="-", marker="o")
ax.plot(x, sp_short_ratio[non_ovip_idx], label="Spermatheca Shorter", color="#AA5440", linestyle="--", marker="s")
ax.plot(x, sp_long_ratio[non_ovip_idx], label="Spermatheca Long", color="#AA5440", linestyle="-", marker="s")

ax.set_xticks(x)
ax.set_xticklabels([time_points[i] for i in non_ovip_idx])
ax.set_title("Non-oviposited")
ax.set_xlabel("Hours post-mating (hr)")
ax.set_ylabel("Sperm proportion")

# ---------- Oviposited ----------
ax = axes[1]
x = np.arange(len(ovip_idx))

ax.plot(x, uterus_short_ratio[ovip_idx], label="Uterus Shorter", color="#385B6F", linestyle="--", marker="o")
ax.plot(x, uterus_long_ratio[ovip_idx], label="Uterus Long", color="#385B6F", linestyle="-", marker="o")
ax.plot(x, sp_short_ratio[ovip_idx], label="Spermatheca Shorter", color="#AA5440", linestyle="--", marker="s")
ax.plot(x, sp_long_ratio[ovip_idx], label="Spermatheca Long", color="#AA5440", linestyle="-", marker="s")

ax.set_xticks(x)
ax.set_xticklabels([time_points[i] for i in ovip_idx])
ax.set_title("Oviposited")
ax.set_xlabel("Hours post-mating (hr)")

# ---------- Legend ----------
fig.legend(loc="right", bbox_to_anchor=(0.5, 1.05), ncol=4)
plt.tight_layout()
#plt.savefig("D:/Documents/PMR/文献/2sperm/2sperm_ratio_change_PM.pdf", dpi=300, bbox_inches='tight', format='pdf')
plt.show()
```
